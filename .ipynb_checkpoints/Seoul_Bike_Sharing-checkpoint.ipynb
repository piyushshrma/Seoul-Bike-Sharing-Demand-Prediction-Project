```python
from statsmodels.stats.outliers_influence import variance_inflation_factor
```

```python
def get_vif(df):
    # Convert all columns to numeric, errors will be set as NaN for non-numeric values
    df = df.apply(pd.to_numeric, errors='coerce')
    
    # Drop rows with NaN values in any column (this removes rows that can't be converted to numbers)
    df = df.dropna()
    
    # Check if the dataframe is empty after cleaning
    if df.empty:
        raise ValueError("Filtered DataFrame for VIF calculation is empty. Check column exclusions.")
    
    # Compute VIF
    vif_data = pd.DataFrame()
    vif_data["variables"] = df.columns
    vif_data["VIF"] = [variance_inflation_factor(df.values, i) for i in range(df.shape[1])]
    
    return vif_data
```

```python
# Drop columns with high VIF or irrelevant columns
df.drop(["Dew point temperature(°C)"], axis=1, inplace=True)
```

```python
# Validate the updated DataFrame
print("\nUpdated DataFrame Info:")
df.info()
```

```python
# Convert boolean columns to uint8
bool_columns = ['Spring', 'Summer', 'Winter', 'Monday', 'Saturday', 'Sunday', 'Thursday', 'Tuesday', 'Wednesday']

# Apply conversion to uint8 for the listed columns
df[bool_columns] = df[bool_columns].astype('uint8')

# Optionally, check the data types to confirm the change
print(df.dtypes)
```

### Encoding

```python
# Manually encode categorical features
df["Holiday"] = df["Holiday"].map({"No Holiday": 0, "Holiday": 1})
df["Functioning Day"] = df["Functioning Day"].map({"No": 0, "Yes": 1})
```

```python
# One-hot encode the 'Seasons' and 'weekday' columns
df_season = pd.get_dummies(df["Seasons"], drop_first=True)
df_weekday = pd.get_dummies(df["weekday"], drop_first=True)
```

```python
# Concatenate the one-hot encoded columns to the main DataFrame
df = pd.concat([df, df_season, df_weekday], axis=1)

# Drop the original 'Seasons' and 'weekday' columns
df.drop(["Seasons", "weekday"], axis=1, inplace=True)
```

```python
# Ensure relevant columns are numeric (if not, convert them)
numeric_columns = ["Rented Bike Count", "Rainfall(mm)", "Snowfall (cm)"]
for col in numeric_columns:
    df[col] = pd.to_numeric(df[col], errors='coerce')

# Handle NaN values by filling with 0 or removing rows with NaN values
df.fillna(0, inplace=True)  # You can also use df.dropna() if you prefer dropping NaNs
```

### Split data for train and test

```python
X = df.drop("Rented Bike Count", axis=1)
y = df["Rented Bike Count"]
```
