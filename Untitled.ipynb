```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import warnings
warnings.filterwarnings('ignore')
from statsmodels.stats.outliers_influence import variance_inflation_factor

# Load dataset
data_path = r"SeoulBikeData.csv"
df = pd.read_csv(data_path, encoding="unicode_escape")

# Convert Date column to datetime
df["Date"] = pd.to_datetime(df["Date"], format="%d/%m/%Y")

# Extract date-related features
df["weekday"] = df["Date"].dt.day_name()
df["Day"] = df["Date"].dt.day
df["Month"] = df["Date"].dt.month
df["Year"] = df["Date"].dt.year

# Drop the original Date column
df.drop("Date", axis=1, inplace=True)

# Manually encode categorical features
df["Holiday"] = df["Holiday"].map({"No Holiday": 0, "Holiday": 1})
df["Functioning Day"] = df["Functioning Day"].map({"No": 0, "Yes": 1})

# One-hot encode the 'Seasons' and 'weekday' columns
df_season = pd.get_dummies(df["Seasons"], drop_first=True)
df_weekday = pd.get_dummies(df["weekday"], drop_first=True)

# Concatenate the one-hot encoded columns to the main DataFrame
df = pd.concat([df, df_season, df_weekday], axis=1)

# Drop the original 'Seasons' and 'weekday' columns
df.drop(["Seasons", "weekday"], axis=1, inplace=True)

# Ensure relevant columns are numeric (if not, convert them)
numeric_columns = ["Rented Bike Count", "Rainfall(mm)", "Snowfall (cm)"]
for col in numeric_columns:
    df[col] = pd.to_numeric(df[col], errors='coerce')

# Handle NaN values by filling with 0 or removing rows with NaN values
df.fillna(0, inplace=True)  # You can also use df.dropna() if you prefer dropping NaNs

# Check the columns after encoding to ensure the correct columns are present
print("\nColumns after encoding:")
print(df.columns)

# Define a random color palette for the plots
random_colors = sns.color_palette("Set1", n_colors=5)

# Barplot by month
plt.figure(figsize=(10,7))
Month = df.groupby("Month").sum().reset_index()
sns.barplot(x="Month", y="Rented Bike Count", data=Month, palette=random_colors)
plt.show()

# Barplot by day
plt.figure(figsize=(10,7))
Day = df.groupby("Day").sum().reset_index()
sns.barplot(x="Day", y="Rented Bike Count", data=Day, palette=random_colors)
plt.show()

# Barplot by hour
plt.figure(figsize=(10,7))
Hour = df.groupby("Hour").sum().reset_index()
sns.barplot(x="Hour", y="Rented Bike Count", data=Hour, palette=random_colors)
plt.show()

# Barplot by holiday
plt.figure(figsize=(10,7))
sns.barplot(x="Holiday", y="Rented Bike Count", data=df, palette=random_colors)
plt.show()

# Barplot for seasons (Spring, Summer, Autumn)
# Check if the columns exist before plotting
seasons_columns = ['Spring', 'Summer', 'Autumn']

for season in seasons_columns:
    if season in df.columns:
        plt.figure(figsize=(10,7))
        sns.barplot(x=season, y="Rented Bike Count", data=df, palette=random_colors)
        plt.title(f"Rented Bike Count by {season}")
        plt.show()
    else:
        print(f"Column '{season}' not found in the dataset, skipping plot.")

# Barplot for rainfall and snowfall
plt.figure(figsize=(10,7))
sns.barplot(x="Rainfall(mm)", y="Rented Bike Count", data=df, palette=random_colors)
plt.show()

plt.figure(figsize=(10,7))
sns.barplot(x="Snowfall (cm)", y="Rented Bike Count", data=df, palette=random_colors)
plt.show()

# Distribution of Rented Bike Count
sns.displot(df["Rented Bike Count"])
plt.show()

# Distribution after applying sqrt transformation
sns.displot(np.sqrt(df["Rented Bike Count"]))
plt.show()

# Check skewness and correlations
print("\nSkewness of features:")
print(df.skew().sort_values(ascending=True))

# Correlation matrix
plt.figure(figsize=(11,11))
sns.heatmap(df.corr(), annot=True, cmap="coolwarm")
plt.show()

def get_vif(df):
    # Convert all columns to numeric, errors will be set as NaN for non-numeric values
    df = df.apply(pd.to_numeric, errors='coerce')
    
    # Drop rows with NaN values in any column (this removes rows that can't be converted to numbers)
    df = df.dropna()
    
    # Check if the dataframe is empty after cleaning
    if df.empty:
        raise ValueError("Filtered DataFrame for VIF calculation is empty. Check column exclusions.")
    
    # Compute VIF
    vif_data = pd.DataFrame()
    vif_data["variables"] = df.columns
    vif_data["VIF"] = [variance_inflation_factor(df.values, i) for i in range(df.shape[1])]
    
    return vif_data

# Drop columns with high VIF or irrelevant columns
df.drop(["Dew point temperature(°C)"], axis=1, inplace=True)

# Validate the updated DataFrame
print("\nUpdated DataFrame Info:")
df.info()
```
